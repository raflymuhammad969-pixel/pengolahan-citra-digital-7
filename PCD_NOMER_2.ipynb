```python
# =======================================
# Soal No. 2 - Konvolusi dengan 3 Kernel
# CPMK 4
# =======================================

import cv2
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Baca citra grayscale ---
# Ganti 'gambar.jpg' dengan nama file kamu (.jpg atau .png)
img = cv2.imread('sate ayam.jpg', cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError("Gambar tidak ditemukan! Pastikan file ada di folder yang sama.")

# --- 2. Definisikan tiga kernel ---
# Low Pass Filter (LPF) - untuk blur / reduksi noise
kernel_LPF = (1/9) * np.array([[1,1,1],
                               [1,1,1],
                               [1,1,1]])

# High Pass Filter (HPF) - untuk deteksi tepi
kernel_HPF = np.array([[-1,-1,-1],
                       [-1, 8,-1],
                       [-1,-1,-1]])

# Band Pass Filter (BPF) - untuk penajaman (sharpening)
kernel_BPF = np.array([[ 0,-1, 0],
                       [-1, 5,-1],
                       [ 0,-1, 0]])

# --- 3. Terapkan filter (konvolusi) pada citra ---
lpf_result = cv2.filter2D(img, -1, kernel_LPF)
hpf_result = cv2.filter2D(img, -1, kernel_HPF)
bpf_result = cv2.filter2D(img, -1, kernel_BPF)

# --- 4. Tampilkan hasil ---
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Citra Asli")
plt.axis('off')

plt.subplot(2, 2, 2)
plt.imshow(lpf_result, cmap='gray')
plt.title("Low Pass Filter (LPF)")
plt.axis('off')

plt.subplot(2, 2, 3)
plt.imshow(hpf_result, cmap='gray')
plt.title("High Pass Filter (HPF)")
plt.axis('off')

plt.subplot(2, 2, 4)
plt.imshow(bpf_result, cmap='gray')
plt.title("Band Pass Filter (BPF)")
plt.axis('off')

plt.tight_layout()
plt.show()

# --- 5. Analisis Nilai Piksel Rata-rata ---
print("=== Nilai Rata-rata Piksel (Mean Intensity) ===")
print(f"Asli : {img.mean():.2f}")
print(f"LPF  : {lpf_result.mean():.2f}")
print(f"HPF  : {hpf_result.mean():.2f}")
print(f"BPF  : {bpf_result.mean():.2f}")
```
